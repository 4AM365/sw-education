# 01: Exploratory Data Analysis

**Objective**: Understand data characteristics, distributions, and patterns.

**Outputs**: Summary statistics, data quality report, visualization plots

```python
# Setup and Configuration
import os
import sys
from dotenv import load_dotenv
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')

# Load environment variables
load_dotenv()

# Set random seeds for reproducibility
SEED = int(os.getenv('RANDOM_SEED', 42))
np.random.seed(SEED)

# Configure visualization
sns.set_style('whitegrid')
plt.rcParams['figure.figsize'] = (12, 6)
plt.rcParams['figure.dpi'] = 100
plt.rcParams['font.size'] = 10

print(f"Random Seed: {SEED}")
print(f"Data Path: {os.getenv('DATA_PATH')}")
```

## Data Loading

```python
# Load data
data_path = os.getenv('DATA_PATH', './data/raw')

# Example: load CSV
# df = pd.read_csv(os.path.join(data_path, 'dataset.csv'))

# For now, create a sample dataset for demonstration
np.random.seed(SEED)
n_samples = 1000
df = pd.DataFrame({
    'feature_1': np.random.normal(100, 15, n_samples),
    'feature_2': np.random.normal(50, 10, n_samples),
    'feature_3': np.random.choice(['A', 'B', 'C'], n_samples),
    'target': np.random.choice([0, 1], n_samples)
})

print(f"Dataset shape: {df.shape}")
print(f"\nFirst few rows:")
print(df.head())
```

## Data Quality Assessment

```python
# Data types and missing values
print("Data Info:")
print(df.info())
print("\nMissing Values:")
print(df.isnull().sum())
print("\nBasic Statistics:")
print(df.describe())
```

## Distribution Analysis

```python
# Visualize distributions
fig, axes = plt.subplots(2, 2, figsize=(12, 10))

# Numeric distributions
axes[0, 0].hist(df['feature_1'], bins=30, edgecolor='black', alpha=0.7)
axes[0, 0].set_title('Feature 1 Distribution')
axes[0, 0].set_xlabel('Value')
axes[0, 0].set_ylabel('Frequency')

axes[0, 1].hist(df['feature_2'], bins=30, edgecolor='black', alpha=0.7)
axes[0, 1].set_title('Feature 2 Distribution')
axes[0, 1].set_xlabel('Value')
axes[0, 1].set_ylabel('Frequency')

# Categorical distribution
feature_3_counts = df['feature_3'].value_counts()
axes[1, 0].bar(feature_3_counts.index, feature_3_counts.values)
axes[1, 0].set_title('Feature 3 (Categorical)')
axes[1, 0].set_ylabel('Count')

# Target distribution
target_counts = df['target'].value_counts()
axes[1, 1].bar(['Class 0', 'Class 1'], target_counts.values)
axes[1, 1].set_title('Target Distribution')
axes[1, 1].set_ylabel('Count')

plt.tight_layout()
plt.savefig(os.path.join(os.getenv('FIGURES_PATH', './results/figures'), '01_distributions.png'), dpi=300, bbox_inches='tight')
plt.show()

print(f"Target balance: {target_counts.to_dict()}")
```

## Correlation Analysis

```python
# Correlation matrix for numeric features
numeric_cols = df.select_dtypes(include=[np.number]).columns
corr_matrix = df[numeric_cols].corr()

plt.figure(figsize=(8, 6))
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', center=0, square=True, linewidths=1)
plt.title('Feature Correlation Matrix')
plt.tight_layout()
plt.savefig(os.path.join(os.getenv('FIGURES_PATH', './results/figures'), '01_correlations.png'), dpi=300, bbox_inches='tight')
plt.show()

print("Correlation with Target:")
print(corr_matrix['target'].sort_values(ascending=False) if 'target' in numeric_cols else "No numeric target")
```
